```python
import kagglehub
import os
import csv
import pandas as pd
```

```python
# Baixar a última versão do dataset Olist
path_olist = kagglehub.dataset_download("olistbr/brazilian-ecommerce")
print("Pasta do dataset:", path_olist)
print(os.listdir(path_olist))
```

```python
order = os.path.join(path_olist, "olist_order_items_dataset.csv")
product = os.path.join (path_olist, "olist_products_dataset.csv")

df_order = pd.read_csv(order)
df_product = pd.read_csv(product)

print("==== Dataframe de ordem de produtos ====")
display(df_order.head(1))
display(df_order.tail(1))

```

```python
df_order.info()
```

```python
df_product.info()
```

```python
print("==== Dataframe dos produtos ====")
display(df_product.head(1))
display(df_product.tail(1))
```

```python
# O merge é com a coluna product_id da product com a product_id da order
df_order_product = pd.merge (
    df_order,
    df_product,
    how="left",
    on="product_id"
)
display(df_order_product.head())
```
